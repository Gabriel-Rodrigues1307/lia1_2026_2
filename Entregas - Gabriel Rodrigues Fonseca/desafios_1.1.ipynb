### 💪 Desafio 1: Análise de Dados de Vendas

```python
# 1. Qual vendedor teve o maior número de vendas?

import numpy as np
import pandas as pd

print(f"NumPy versão: {np.__version__}")
print(f"Pandas versão: {pd.__version__}")

# Criando o DataFrame df_vendas_completo
np.random.seed(42) # Para reprodutibilidade

dados_vendas = {
    'Vendedor': ['Ana', 'João', 'Pedro', 'Ana', 'João', 'Pedro', 'Ana', 'João', 'Pedro', 'Ana'] * 3,
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'] * 6,
    'Regiao': ['Norte', 'Sul', 'Nordeste', 'Centro-Oeste', 'Sudeste'] * 6,
    'Vendas': np.random.randint(1000, 10000, 30),
    'Quantidade': np.random.randint(1, 20, 30)
}

df_vendas_completo = pd.DataFrame(dados_vendas)

vendas_por_vendedor = df_vendas_completo.groupby('Vendedor')['Vendas'].sum()
print("Vendas totais por vendedor:")
print(vendas_por_vendedor)
maior_vendedor = vendas_por_vendedor.idxmax()
print(f"\nO vendedor com o maior número de vendas foi: {maior_vendedor} com {vendas_por_vendedor.max():.2f}")
```

```python
# 2. Qual região teve o maior faturamento total?
vendas_por_regiao = df_vendas_completo.groupby('Regiao')['Vendas'].sum()
print("\nFaturamento total por região:")
print(vendas_por_regiao)
maior_regiao = vendas_por_regiao.idxmax()
print(f"\nA região com o maior faturamento total foi: {maior_regiao} com {vendas_por_regiao.max():.2f}")
```

```python
# 3. Qual é o produto mais vendido em quantidade?
quantidade_por_produto = df_vendas_completo.groupby('Produto')['Quantidade'].sum()
print("\nQuantidade vendida por produto:")
print(quantidade_por_produto)
produto_mais_vendido = quantidade_por_produto.idxmax()
print(f"\nO produto mais vendido em quantidade foi: {produto_mais_vendido} com {quantidade_por_produto.max()} unidades")
```

```python
# 4. Qual é a média de vendas por vendedor?
media_vendas_por_vendedor = df_vendas_completo.groupby('Vendedor')['Vendas'].mean()
print("\nMédia de vendas por vendedor:")
print(media_vendas_por_vendedor)
```

### 🔧 Desafio 2: Limpeza de Dados

```python
import numpy as np
import pandas as pd

# Criando um DataFrame com dados
```

```python
# Criando um DataFrame com dados "problemáticos"
dados_problematicos = {
    'ID_Transacao': [1, 2, 3, 1, 4, 5, 6, 7],
    'Produto': ['  Livro  ', 'Caneta', 'Lapis', 'Livro', 'Borracha', 'Apontador', 'Caderno', '  CADERNO '],
    'Preco': [15.50, 2.00, None, 15.50, 3.50, 1.80, 25.00, 25.00],
    'Quantidade': [2, 1, 5, 2, 3, None, 1, 1],
    'Categoria': ['LIVROS', 'Escrita', 'Desenho', 'LIVROS', 'Escrita', 'Escrita', 'Papelaria', 'PAPELARIA'],
    'Desconto': [0.1, 0.0, 0.05, 0.1, 0.0, 0.0, None, 0.0]
}

df_problema = pd.DataFrame(dados_problematicos)
print("DataFrame Original com Problemas:")
display(df_problema)

df_limpeza_desafio = df_problema.copy()

# 1. Remoção de duplicatas
print("\n--- Após remoção de duplicatas ---")
df_limpeza_desafio.drop_duplicates(inplace=True)
display(df_limpeza_desafio)

# 2. Tratamento de valores ausentes
# Preencher Preco com a média, Quantidade com a mediana e Desconto com 0
df_limpeza_desafio['Preco'].fillna(df_limpeza_desafio['Preco'].mean(), inplace=True)
df_limpeza_desafio['Quantidade'].fillna(df_limpeza_desafio['Quantidade'].median(), inplace=True)
df_limpeza_desafio['Desconto'].fillna(0, inplace=True)
print("\n--- Após tratamento de valores ausentes ---")
display(df_limpeza_desafio)

# 3. Padronização de strings
df_limpeza_desafio['Produto'] = df_limpeza_desafio['Produto'].str.strip().str.title()
df_limpeza_desafio['Categoria'] = df_limpeza_desafio['Categoria'].str.strip().str.capitalize()
print("\n--- Após padronização de strings ---")
display(df_limpeza_desafio)

# 4. Criação de novas colunas derivadas
df_limpeza_desafio['Valor_Total'] = df_limpeza_desafio['Preco'] * df_limpeza_desafio['Quantidade'] * (1 - df_limpeza_desafio['Desconto'])
df_limpeza_desafio['É_Descontado'] = np.where(df_limpeza_desafio['Desconto'] > 0, 'Sim', 'Não')
print("\n--- Após criação de novas colunas ---")
display(df_limpeza_desafio)
```

### 📊 Desafio 3: Visualização

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

# 1. Comparação entre vendedores
vendas_por_vendedor = df_vendas_completo.groupby('Vendedor')['Vendas'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=vendas_por_vendedor.index, y=vendas_por_vendedor.values, palette='viridis')
plt.title('Total de Vendas por Vendedor')
plt.xlabel('Vendedor')
plt.ylabel('Total de Vendas')
plt.show()
```

```python
# 2. Tendências por região
vendas_por_regiao = df_vendas_completo.groupby('Regiao')['Vendas'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(x=vendas_por_regiao.index, y=vendas_por_regiao.values, palette='magma')
plt.title('Faturamento Total por Região')
plt.xlabel('Região')
plt.ylabel('Faturamento Total')
plt.show()
```

```python
# 3. Distribuição dos valores (Vendas)
plt.figure(figsize=(10, 6))
sns.histplot(df_vendas_completo['Vendas'], bins=15, kde=True, color='skyblue')
plt.title('Distribuição dos Valores de Vendas')
plt.xlabel('Valores de Vendas')
plt.ylabel('Frequência')
plt.show()

# Distribuição dos valores (Quantidade)
plt.figure(figsize=(10, 6))
sns.histplot(df_vendas_completo['Quantidade'], bins=10, kde=True, color='lightcoral')
plt.title('Distribuição dos Valores de Quantidade')
plt.xlabel('Valores de Quantidade')
plt.ylabel('Frequência')
plt.show()
```
